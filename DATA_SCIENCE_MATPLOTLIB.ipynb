```python
import matplotlib.pyplot as plt #pyplot is used to add extra features to matplotlib   #matplotlib(tool) is used for data visualisation
x = [1,2,300,4,500]
y = [2,4,600,80,10]

plt.plot(x,y)  #plot is used to plot the x and y values in graph
plt.show()  #show is used to display the graph
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 20, 60, 100]
plt.plot(months,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 20, 60, 100]
profit = [1,23,4,5,76]

plt.plot(months,sales,label="sales")  #label - used to give names on graph- like sales, profit etc....
plt.plot(months,profit,label="profit")
plt.legend()  #used to display the label names - if legend is not used the names will not be displayed

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 20, 60, 100]

plt.plot(months,sales)

plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.grid(True)
plt.plot(months,sales,marker="*",color ="green",linestyle = "--")  #marker="o" is used show that dot in graph, marker = "s" shows square, msrker = "*" shows * on graph
#linestyle is used to show the line in graph
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 20, 60, 100]
profit = [1,23,4,5,76]

plt.bar(months,sales,label="sales") #.bar is used to display the graph in bar format
plt.bar(months,profit,label="profit")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1,2,3,4,5,6]
marks = [40,45,55,65,75,85]

plt.scatter(study_hours, marks)  #scatter is used to show the dots(points) in graph
plt.title("study hours vs marks")
plt.xlabel("study hours")
plt.ylabel("marks")

plt.show()
```

```python
import matplotlib.pyplot as plt
marks = [45,67,89,90,53,60]
plt.hist(marks)  #hist is used to represent histogram
plt.title('marks distribution')
plt.xlabel('marks')
plt.ylabel('frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
marks = [45,67,89,90,53,6,100,23,56,90,63,47,52,65]
plt.pie(marks)  #pie is used to represent piechart
plt.title('marks distribution')
plt.xlabel('marks')
plt.ylabel('frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax = plt.subplots(2,1)  #ax means indivudual graphs and fig means whole graph, subplots means plots under another plots
ax[0].plot([1,2,3],[10,20,30])  #ax[0] means 1st graph representation in the form of graph
ax[1].bar(["A","B","C"],[20,15,30]) #ax[1] means to represent second one as bar graph
plt.show()
```
